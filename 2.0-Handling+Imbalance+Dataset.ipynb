## Handling Imbalanced Dataset

1. Up Sampling
2. Down Sampling

```python
import numpy as np
import pandas as pd

# Set the random seed for reproducibility
np.random.seed(123)

# Create a dataframe with two classes
n_samples = 1000
class_0_ratio = 0.9
n_class_0 = int(n_samples * class_0_ratio)
n_class_1 = n_samples - n_class_0
```

```python
n_class_0,n_class_1
```

```python
## CREATING MY DATAFRAME WITH IMBALACED DATASET

class_0 = pd.DataFrame({
    'feature _1' : np.random.normal(loc = 0,scale = 1,size = n_class_0),
    'feature_2' : np.random.normal(loc = 0,scale = 1,size = n_class_0),
    'target' : [0] * n_class_0
})
class_1 = pd.DataFrame({
    'feature _1' : np.random.normal(loc = 0,scale = 1,size = n_class_1),
    'feature_2' : np.random.normal(loc = 0,scale = 1,size = n_class_1),
    'target' : [1] * n_class_1
})
```

```python
df = pd.concat([class_1,class_0]).reset_index(drop = True)
```

```python
df['target'].value_counts()
```

```python
df_min = df[df['target']==1]
df_maj = df[df['target']==0]
```

```python
from sklearn.utils import resample
df_min_upsampled = resample(df_min,replace=True,n_samples=len(df_maj),random_state=42)
df_min_upsampled.shape
```

```python
df_min_upsampled.head()
```

```python
df_upsampled = pd.concat([df_maj,df_min_upsampled])
df_upsampled['target'].value_counts()
```

## Down Sampling

Here we use the same data sets for down sampling also

```python
from sklearn.utils import resample
df_maj_downsampled = resample(df_maj,replace=False,n_samples=len(df_maj),random_state=42)
df_maj_downsampled.shape
```

```python
df_downsampled = pd.concat([df_min,df_maj_downsampled])
df_downsampled['target'].value_counts()
```

## Disadvantages of Upsampling
1) Severe Risk of Overfitting: Because standard upsampling simply duplicates existing rows, the machine learning model evaluates the exact same minority data points repeatedly.
2) Increased Computational Cost: Expanding the minority class to match the majority class drastically increases the total size of your dataset. This requires more RAM to hold the DataFrame in memory
3) Amplification of Noise: If there are outliers, errors, or noisy data points within your small minority class, upsampling will duplicate those errors, forcing the model to treat mere anomalies as significant, repeating trends.

## Disadvantages of Downsampling
1) Severe Information Loss: By randomly deleting rows from the majority class to match the minority class, you actively throw away valid, potentially crucial data.
2) Underfitting the Majority Class: Because the model is trained on a drastically reduced version of the majority class, it fails to learn the full complexity, edge cases, and natural variance of that specific group.
3) Unrepresentative Sampling: The random subset of the majority class that survives the downsampling process might not accurately represent the true distribution of the original dataset.
